# break and continue



```python
lst=['ram', 'ROHIT', 'SNEHA', 2, 4]

for name in lst:
    print(name)
```

```python
for name in lst:
    if name=='ram':
        print(f"name found: {name}")
        break
        print(f"my name is {name}")
```

```python
for name in lst:
    if name=='ram':
        print(f"name found: {name}")
        continue
        print(f"my name is {name}")
```

```python
def add_num(a,b):
    '''This function takes two arguments and returns their sum, difference, product, and quotient.'''
    sum=a+b
    subs=a-b
    multiply=a*b
    divide=a/b  
    return sum, subs, multiply, divide
```

# recursion : Function call itself

```python
factorial=1
for i in range(1,6):
    factorial=factorial*i
print(factorial)
```

```python
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact
```

```python
def rec_fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*rec_fact(n-1)
```
